```python
import pandas as pd
import numpy as np

# --- Célula Única: Tabela de Métricas por Campo (Versão Robusta) ---

file_path = 'relatorio_metricas_por_arquivo.csv'

try:
    # 1. Carregar o CSV
    # Usamos 'dtype=str' para ler tudo como texto inicialmente e evitar confusão do pandas
    df = pd.read_csv(file_path, sep=';', dtype=str)
    
    # Colunas que DEVEM ser numéricas
    cols_metrics = ['Acc_Nome', 'Acc_Lugar', 'Acc_DataIni', 'Acc_DataFim', 'Acc_Desc', 'F1-Score']
    
    # 2. Tratamento de Dados (Conversão Forçada)
    for col in cols_metrics:
        if col in df.columns:
            # a) Remove '%' e troca ',' por '.'
            # b) Usa to_numeric com errors='coerce' (transforma erro em NaN)
            # c) Preenche NaN com 0.0
            df[col] = pd.to_numeric(
                df[col].str.replace('%', '').str.replace(',', '.'), 
                errors='coerce'
            ).fillna(0.0)

    # 3. Seleção e Renomeação
    view_cols = ['Arquivo', 'Acc_Nome', 'Acc_Lugar', 'Acc_DataIni', 'Acc_DataFim', 'Acc_Desc', 'F1-Score']
    
    # Verifica se todas as colunas existem antes de selecionar
    cols_existentes = [c for c in view_cols if c in df.columns]
    df_view = df[cols_existentes].copy()
    
    # Dicionário de renomeação (apenas para as colunas que existem)
    rename_map = {
        'Arquivo': 'Arquivo',
        'Acc_Nome': 'Nome (%)',
        'Acc_Lugar': 'Lugar (%)',
        'Acc_DataIni': 'Data Início (%)',
        'Acc_DataFim': 'Data Fim (%)',
        'Acc_Desc': 'Descrição (%)',
        'F1-Score': 'F1-Geral (%)'
    }
    df_view = df_view.rename(columns=rename_map)
    
    # Identificar as colunas numéricas finais para o estilo
    # (são todas menos 'Arquivo')
    numeric_cols = [c for c in df_view.columns if c != 'Arquivo']

    # 4. Exibição com Estilo
    print("Tabela de Métricas por Campo (Arquivo por Arquivo):")
    
    # Aplica o gradiente e a formatação APENAS nas colunas numéricas
    tabela_estilizada = df_view.style.background_gradient(
        cmap='RdYlGn', 
        vmin=0, 
        vmax=100, 
        subset=numeric_cols
    ).format("{:.2f}", subset=numeric_cols)

    display(tabela_estilizada)

except FileNotFoundError:
    print(f"ERRO: O arquivo '{file_path}' não foi encontrado. Certifique-se de ter rodado o script de validação.")
except Exception as e:
    print(f"ERRO ao processar dados: {e}")
    # Se der erro, imprime o dataframe bruto para debug
    if 'df' in locals():
        print("Dados brutos lidos (primeiras linhas):")
        display(df.head(20))
```
